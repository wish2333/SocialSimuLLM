# Quick Start: Loading and Exploring Experiment Data

This notebook demonstrates how to load experiment results and perform
basic exploration of agent behavior in SocialSimuLLM.

**Prerequisites**: Run an experiment first (see `examples/python_runner.py`)
to generate data under `runs/<experiment_id>/`.

```python
from data_loader import (
    load_experiment,
    load_checkpoints,
    build_agent_timeline,
    extract_action_events,
    extract_movement_events,
    extract_reflection_events,
)

import json
from collections import Counter
```

## 1. Load Experiment Data

Replace `"exp_abc123"` with your actual experiment ID.

```python
experiment_id = "exp_abc123"  # <-- change this
events = load_experiment(experiment_id)
print(f"Loaded {len(events)} events")
```

## 2. Explore Event Structure

Each event is a dict with keys: `timestamp`, `step`, `agent_id`, `event_type`, `data`.

```python
# Show first 3 events
for e in events[:3]:
    print(json.dumps(e, indent=2, ensure_ascii=False))
    print("---")
```

```python
# Event type distribution
type_counts = Counter(e.get("event_type", "unknown") for e in events)
print("Event type distribution:")
for etype, count in type_counts.most_common():
    print(f"  {etype}: {count}")
```

## 3. Filter Events by Type

```python
actions = extract_action_events(events)
movements = extract_movement_events(events)
reflections = extract_reflection_events(events)

print(f"Actions: {len(actions)}")
print(f"Movements: {len(movements)}")
print(f"Reflections: {len(reflections)}")
```

## 4. Build Agent Timeline

Filter all events for a specific agent to see their individual story.

```python
# List all agents
agents = sorted(set(e.get("agent_id", "") for e in events if e.get("agent_id")))
print("Available agents:", agents)
```

```python
agent_name = agents[0] if agents else "unknown"
timeline = build_agent_timeline(events, agent_name)
print(f"{agent_name} has {len(timeline)} events")
print()
for e in timeline[:5]:
    print(f"  Step {e.get('step')}: [{e.get('event_type')}] {str(e.get('data', ''))[:100]}")
```

## 5. Load Checkpoints

Checkpoints store agent states and spatial graph at periodic intervals.

```python
checkpoints = load_checkpoints(experiment_id)
print(f"Loaded {len(checkpoints)} checkpoints")
if checkpoints:
    cp = checkpoints[0]
    print(f"First checkpoint keys: {list(cp.keys())}")
    if "agent_states" in cp:
        print(f"Agents in checkpoint: {list(cp['agent_states'].keys())[:5]}")
```
